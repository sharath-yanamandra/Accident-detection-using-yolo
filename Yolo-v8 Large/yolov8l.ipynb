```python
import os
from ultralytics import YOLO

# Defined the function to check accidents
def check_acc(box):
    res_index_list = box.cls.tolist()
    result = ""

    for index in res_index_list:
        if index == 1:
            result = "Bike Bike Accident Detected"
            break
        elif index == 2:
            result = "Bike Object Accident Detected"
            break
        elif index == 3:
            result = "Bike Person Accident Detected"
            break
        elif index == 5:
            result = "Car Bike Accident Detected"
            break
        elif index == 6:
            result = "Car Car Accident Detected"
            break
        elif index == 7:
            result = "Car Object Accident Detected"
            break
        elif index == 8:
            result = "Car Person Accident Detected"
            break
    
    return result


# Path to the folders containing test images
accident_folder_path = "/kaggle/input/dataset/data/test/images/Accident"
non_accident_folder_path = "/kaggle/input/dataset/data/test/images/Non Accident"

# Initialize YOLO model
model = YOLO("/kaggle/working/runs/detect/train/weights/best.pt")

# Initialize variables for counting
total_accident_images = 0
accident_detected_count = 0

# Iterate over each image in the accident folder
for filename in os.listdir(accident_folder_path):
    if filename.endswith(".jpg") or filename.endswith(".png"):
        res = ""
        total_accident_images += 1
        image_path = os.path.join(accident_folder_path, filename)
        
        # Predict using YOLO model
        results = model.predict(source=image_path)
        
        # Check for accident detection
        box = results[0].boxes
        res = check_acc(box)

        # If accident detected, increase the count
        if len(res) > 0:
            accident_detected_count += 1

# Calculate accuracy for accident images
accident_accuracy = (accident_detected_count / total_accident_images) * 100 if total_accident_images != 0 else 0

print("\nTotal accident images processed:", total_accident_images)
print("Accident detected in", accident_detected_count, "accident images")
print("Accident Detection Accuracy:", accident_accuracy, "%\n")

# Initialize variables for counting non-accident images
total_non_accident_images = 0
non_accident_detected_count = 0

# Iterate over each image in the non-accident folder
for filename in os.listdir(non_accident_folder_path):
    if filename.endswith(".jpg") or filename.endswith(".png"):
        res = ""
        total_non_accident_images += 1
        image_path = os.path.join(non_accident_folder_path, filename)
        
        # Predict using YOLO model
        results = model.predict(source=image_path)
        
        # Check for accident detection
        box = results[0].boxes
        res = check_acc(box)
        
        if len(res) == 0:
            non_accident_detected_count += 1

# Calculate accuracy for non-accident images
non_accident_accuracy = (non_accident_detected_count / total_non_accident_images) * 100 if total_non_accident_images != 0 else 0

print("\nTotal non-accident images processed:", total_non_accident_images)
print("Non-accident detected in", non_accident_detected_count, "non-accident images")
print("Non-Accident Detection Accuracy:", non_accident_accuracy, "%")
```

```python
# Calculate total accuracy
total_images = total_accident_images + total_non_accident_images
total_correct_predictions = accident_detected_count + non_accident_detected_count

total_accuracy = (total_correct_predictions / total_images) * 100 if total_images != 0 else 0

print("\nTotal images processed:", total_images)
print("Total correct predictions:", total_correct_predictions)
print("Overall Accuracy:", total_accuracy, "%")
```

```python
# if conf = x, we will see only those object confidence >= x
import cv2
import time
import matplotlib.pyplot as plt

def extract_result(video):
    vidcap = cv2.VideoCapture(video)
    fps = vidcap.get(cv2.CAP_PROP_FPS)
    nof = 2
    frame_no = 0
    while vidcap.isOpened():
        res = ""
        success, image = vidcap.read()
        if success == False:
            break

        # Check if it's time to process the frame based on the desired interval
        if (frame_no % (int(fps / nof))) == 0:
            results = model.predict(image,conf = 0.6,iou = 0.3,imgsz = 512)  
            box = results[0].boxes
            res = check_acc(box) 
            annotated_frame = results[0].plot()

            # Display the annotated frame only if an accident is detected
            if len(res) > 0:
                print("Accident Type: ", res)
                plt.imshow(cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB))
                plt.axis('off')  # Hide axis
                plt.show()

        frame_no += 1  # Increment frame number

    # Release the video capture object
    vidcap.release()
```

```python
video_path = "/kaggle/input/dataset/data/testing1.mp4"
extract_result(video_path)
```
